```python
import matplotlib.pyplot as plt
import tensorflow as tf
import seaborn as sns
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_circles
```

```python
X, y = make_circles(n_samples=100, noise=0.1, random_state=1)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
X.shape
```

```python
sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=y, palette='Set1')
```

```python
model = Sequential()
```

```python
model.add(Dense(256, input_dim=2, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
history = model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=500, verbose=1)
```

```python
plt.plot(history.history['loss'], label='train')
plt.plot(history.history['val_loss'], label='test')
plt.legend()
plt.show()
```

```python
model = Sequential()
```

```python
model.add(Dense(256, input_dim=2, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
callback = EarlyStopping(monitor='val_loss', min_delta=0.00001, verbose=1, patience=20, mode='auto', restore_best_weights=False)
```

```python
history = model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=500, verbose=1, callbacks=[callback])
```

```python
plt.plot(history.history['loss'], label='train')
plt.plot(history.history['val_loss'], label='test')
plt.legend()
plt.show()
```

```python
my_model = Sequential()
my_model.add(Dense(256, input_dim=2, activation='relu'))
my_model.add(Dense(1, activation='sigmoid'))
```

```python
my_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
my_callback = EarlyStopping(monitor='val_loss', min_delta=0.00001, verbose=1, patience=800, mode='auto', restore_best_weights=False)
```

```python
my_history = my_model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=1000, verbose=1, callbacks=[my_callback])
```
